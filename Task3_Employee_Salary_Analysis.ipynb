# Task 3: Employee Salary Analysis

This notebook analyses a manually created employee salary dataset. It calculates the total, average, highest and lowest salary, and checks the results against the original data.

```
# Manually created employee salary data
employees = [
    {"name": "Asha", "department": "Sales", "salary": 42000},
    {"name": "Ben", "department": "Sales", "salary": 38500},
    {"name": "Chitra", "department": "HR", "salary": 51000},
    {"name": "Dev", "department": "IT", "salary": 67000},
    {"name": "Esha", "department": "IT", "salary": 72000},
    {"name": "Farhan", "department": "Finance", "salary": 58000},
    {"name": "Gita", "department": "HR", "salary": 46500},
    {"name": "Hari", "department": "Finance", "salary": 61000},
]
salaries = [e["salary"] for e in employees]
print("Original data:", salaries)
```

## Salary statistics

The calculations use built-in Python functions: `sum`, `len`, `max` and `min`.

```
total_salary = sum(salaries)
average_salary = total_salary / len(salaries)
highest_salary = max(salaries)
lowest_salary = min(salaries)

print(f"Total salary:   {total_salary:,}")
print(f"Average salary: {average_salary:,.2f}")
print(f"Highest salary: {highest_salary:,}")
print(f"Lowest salary:  {lowest_salary:,}")
```

## Check against the original data

Each result is compared with the original list, so the calculations are verified rather than assumed.

```
assert total_salary == 42000 + 38500 + 51000 + 67000 + 72000 + 58000 + 46500 + 61000
assert highest_salary in salaries and lowest_salary in salaries
assert lowest_salary <= average_salary <= highest_salary
assert abs(average_salary * len(salaries) - total_salary) < 1e-6
print("All checks passed: totals, extremes and the average agree with the original data.")
```

## Answers to the interview questions

1. **How do you calculate the average salary?** Divide the total salary by the number of employees: `total / len(salaries)`.
2. **Difference between total and average salary?** The total is the sum of all salaries. The average is the total divided by the number of employees, so it describes a typical salary.
3. **How would you analyse salary data with Pandas?** Load the data into a `DataFrame`, then use `df['salary'].sum()`, `.mean()`, `.max()`, `.min()`, or `df.groupby('department')['salary'].mean()` for department-level results.
